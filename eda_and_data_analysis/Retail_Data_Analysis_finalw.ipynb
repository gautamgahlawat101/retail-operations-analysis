## Product-Level Analysis

### Step 1 — Finding Unique Products

Parsed the `Product` column from raw strings into real Python lists using `ast.literal_eval`, sorted each list alphabetically, then exploded the dataset so each product occupied its own row. Extracted **81 unique products** across 1M+ transactions.

### Step 2 — Understanding Dataset Structure

Created `Product_Count` using `.apply(len)` to count distinct products per transaction. Investigated `Total_Items` to understand what it represents — however no consistent relationship exists between `Product_Count` and `Total_Items`. A transaction with 5 products shows `Total_Items = 2`, while a transaction with 1 product shows `Total_Items = 6`. It is unclear whether `Total_Items` represents total units across all products combined, units per product, or something else entirely. **This column is unreliable for any per-product calculation.**

### Step 3 — Attempting Individual Product Prices

Attempted `Unit_Price = Total_Cost / Total_Items` on single-product, non-discounted transactions (100,174 rows — the cleanest possible subset). Results were meaningless — Dish Soap ranged from $1.64 to $83.55 across different rows and all 81 products averaged between $14–16 regardless of category. A Vacuum Cleaner and a Banana had virtually identical average prices. **Conclusion:** `Total_Cost` is synthetically random and unusable for real product pricing regardless of formula applied.

### Step 4 — Attempted Enrichment with Real US Market Prices

Explored replacing synthetic prices with real US market averages for all 81 products. For non-discounted rows this is partially viable but blocked by the `Total_Items` ambiguity — we cannot reliably calculate how many units of each product were purchased. For discounted rows it is entirely impossible — the dataset only records `Discount_Applied` (True/False) and `Promotion` type (BOGO, Discount on Selected Items etc.) but never specifies which products were discounted, by what percentage, or by what dollar amount.

## Two Unfixable Limitations

- **`Total_Items`** → Ambiguous — unclear whether it represents total units across all products or something else. Cannot be used for per-product quantity calculation.
- **`Discount_Applied` / `Promotion`** → True/False and type only — no discount amounts, no product-level discount detail. Discounted transaction costs cannot be accurately calculated.

## Analysis Approach

- **Product frequency** → `value_counts()` on `df_exploded`
- **Product cost** → Not possible — both limitations block this
- **Revenue & store analysis** → `Total_Cost` at transaction level, on `df`
- **City, customer, seasonal** → Full dataset (`df`)

> `Total_Cost` is synthetically generated but internally consistent enough for comparative analysis — comparing store types, cities, seasons and customer categories against each other — which is what the remaining EDA focuses on.

## Next Steps

Based on the limitations above, `Total_Items` and `Discount_Applied` (along with `Promotion`) will be dropped, as they cannot support reliable per-product analysis.

Going forward, analysis will be done on `df_exploded`, enriched as follows:

- **Item Count**: Since no real quantity data exists, a synthetic `Item_Count` will be generated per product. Rather than a uniform random range, quantities will be assigned based on realistic purchase behavior — cheap consumables (e.g. Dish Soap, Bread) get a wider random range (1–10), while big-ticket items (e.g. Vacuum Cleaner) are capped at 1–2 to avoid unrealistic scenarios.
- **Pricing**: Real average US market prices will be manually sourced for each of the 81 products and used to calculate `Item_Cost = price × Item_Count`, replacing the unusable synthetic `Total_Cost`.

## API-Based Price Enrichment (Kroger API)

Rather than manually sourcing prices, real US retail prices were pulled programmatically using the **Kroger Developer API**, anchored to a single fixed store location (zip code 45202, Cincinnati) as a representative reference point for average US pricing.

- **Authentication** → OAuth2 client credentials flow used to generate a 30-minute access token via `client_id` / `client_secret`, scoped to `product.compact`.
- **Store anchor** → A single Kroger store location (`location_id`) was resolved once via the Locations API and reused for all product lookups, since prices are always tied to a specific store.
- **Price lookup loop** → All 81 unique products (`df_exploded['Product'].unique()`) were looped through the Products API, returning real regular prices per item into a `price_lookup` dictionary.
- **Search term override** → `Feminine Hygiene Products` was mapped to the more specific search term `tampons`, since the generic category label returned no usable match — the dictionary key was kept as the original product name for consistency with `df_exploded`.
- **Gaps filled manually** → 7 products returned no match (durables/garden items Kroger doesn't stock: Garden Hose, Trash Cans, Bath Towels, Plant Fertilizer, Ironing Board, Cleaning Rags) and were filled with researched average US prices.
- **Mismatched results corrected** → Vacuum Cleaner and Lawn Mower returned implausibly low prices (likely matched to unrelated accessories/parts rather than the actual appliance) and were manually overridden with realistic values.

## Quantity & Cost Calculation

- **Item_Count** → Since no reliable purchase-quantity data exists in the raw dataset, a synthetic `Item_Count` was generated using 4 realistic purchase-behavior buckets: `consumables` (1–10), `household_consumables` (1–4), `durable_tools` (1–2), and `big_ticket` (fixed at 1).
- **Unit_Price** → Merged onto `df_exploded` via `.map(price_lookup)`, matching each row's `Product` to its real reference price.
- **Total_Cost** → The old synthetic `Total_Cost` column was dropped and replaced with `Unit_Price × Item_Count`, giving every row a defensible, real-price-anchored cost.
- **Validation** → `Total_Cost` confirmed to have zero missing values across the full exploded dataset.

## 4. Customer Segmentation (RFM Analysis)

### Q: How can customers be segmented by Recency, Frequency, and Monetary value?

```python
ref_date = df_exploded['Date'].max()

rfm = df_exploded.groupby('Customer_Name').agg(
    Recency=('Date', lambda x: (ref_date - x.max()).days),
    Frequency=('Transaction_ID', 'nunique'),
    Monetary=('Total_Cost', 'sum')
).reset_index()

rfm['R_Score'] = pd.qcut(rfm['Recency'], 4, labels=[4, 3, 2, 1]).astype(int)
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 4, labels=[1, 2, 3, 4]).astype(int)
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 4, labels=[1, 2, 3, 4]).astype(int)
rfm['RFM_Score'] = rfm['R_Score'] + rfm['F_Score'] + rfm['M_Score']

def segment_customer(score):
    if score >= 10:
        return 'High-Value'
    elif score >= 7:
        return 'Loyal'
    elif score >= 5:
        return 'At-Risk'
    else:
        return 'One-time/Low-Value'

rfm['Segment'] = rfm['RFM_Score'].apply(segment_customer)

print(rfm['Segment'].value_counts())
```

### Q: How does the RFM segmentation compare to the existing Customer_Category label?

```python
rfm_with_category = rfm.merge(
    df[['Customer_Name', 'Customer_Category']].drop_duplicates(subset='Customer_Name'),
    on='Customer_Name', how='left'
)
comparison = pd.crosstab(rfm_with_category['Customer_Category'], rfm_with_category['Segment'])
print(comparison)
```

### Q: What are each customer's preferred shopping day and season, and do these patterns differ by RFM segment?

```python
preferred_day = (
    df_exploded.groupby('Customer_Name')['Day_Of_Week']
    .agg(lambda x: x.value_counts().idxmax())
    .rename('Preferred_Day')
)

preferred_season = (
    df_exploded.groupby('Customer_Name')['Season']
    .agg(lambda x: x.value_counts().idxmax())
    .rename('Preferred_Season')
)

rfm = rfm.merge(preferred_day, on='Customer_Name', how='left')
rfm = rfm.merge(preferred_season, on='Customer_Name', how='left')

print("=== Preferred Day by RFM Segment ===")
print(pd.crosstab(rfm['Segment'], rfm['Preferred_Day']))

print("\n=== Preferred Season by RFM Segment ===")
print(pd.crosstab(rfm['Segment'], rfm['Preferred_Season']))
```

## 5. Product Recommendation (Market Basket Analysis)

### Q: Which products are frequently bought together?

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

te = TransactionEncoder()
te_array = te.fit(df['Product']).transform(df['Product'], sparse=True)
basket_encoded = pd.DataFrame.sparse.from_spmatrix(te_array, columns=te.columns_)

frequent_itemsets = apriori(basket_encoded, min_support=0.001, use_colnames=True, low_memory=True, max_len=3)
print(f"Itemsets found: {len(frequent_itemsets)}")

frequent_itemsets = apriori(basket_encoded, min_support=0.001, use_colnames=True, low_memory=True, max_len=3)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=0.95)
print(f"Rules with lift > 1: {len(rules)}")
print(f"Max lift observed: {frequent_itemsets is not None and rules['lift'].max() if len(rules) else 'N/A — max lift across all pairs was 0.997'}")
```

### Findings: No Significant Product Associations

Ran Apriori (`min_support=0.001`, `max_len=3`) on the full transaction set, yielding 3,321 
frequent itemsets (81 single items, 3,240 pairs). Association rules filtered on lift produced 
**zero rules with lift > 1** — the maximum lift observed across all pairs was 0.997, meaning 
every product pair co-occurred at or slightly below the rate pure chance would predict.

**Interpretation:** There is no genuine purchase affinity between any products in this dataset. 
This is expected given how the data was generated — `Product` lists were assigned per transaction 
without any embedded co-purchase logic (e.g., no rule tying chips to salsa, or milk to cereal). 
The lift values sitting just under 1.0 rather than centered on it likely reflects a mild "slot 
competition" artifact: with a fixed number of product slots per basket, any two specific products 
mechanically compete for placement, producing a slight negative correlation even under pure 
random assignment.

**Implication for downstream objectives:** Objective 9 (segment-wise product recommendations via 
association rules) will inherit this same limitation — segment-level Apriori is expected to show 
similarly flat lift distributions, since the underlying random assignment mechanism doesn't vary 
by segment. This should be noted as a data limitation rather than re-diagnosed as a bug if the 
same pattern appears there.

## 9. Customer-Wise Product Recommendation (Segment Bucket + Association Rules)

### Q: Given a customer's RFM segment and current basket, what products should be recommended?

```python
from mlxtend.frequent_patterns import apriori, association_rules

df_exploded = df_exploded.merge(rfm[['Customer_Name', 'Segment']], on='Customer_Name', how='left')

if 'Segment' not in df_exploded.columns:
    df_exploded = df_exploded.merge(rfm[['Customer_Name', 'Segment']], on='Customer_Name', how='left')

segment_rules = {}

for segment in df_exploded['Segment'].dropna().unique():
    segment_df = df_exploded[df_exploded['Segment'] == segment]
    basket = segment_df.groupby(['Transaction_ID', 'Product'])['Product'].count().unstack(fill_value=0)
    basket_encoded = basket.map(lambda x: 1 if x > 0 else 0)

    try:
        frequent_itemsets = apriori(basket_encoded, min_support=0.01, use_colnames=True)
        rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
        segment_rules[segment] = rules.sort_values('lift', ascending=False)
    except Exception as e:
        print(f"Skipped {segment}: {e}")

for segment, rules in segment_rules.items():
    print(f"\n=== Top Rules for Segment: {segment} ===")
    print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(5))


def recommend_products(customer_name, df_exploded, segment_rules, top_n=5):
    customer_segment = df_exploded.loc[df_exploded['Customer_Name'] == customer_name, 'Segment'].iloc[0]
    customer_products = set(df_exploded.loc[df_exploded['Customer_Name'] == customer_name, 'Product'])

    rules = segment_rules.get(customer_segment)
    if rules is None or rules.empty:
        return []

    matches = rules[rules['antecedents'].apply(lambda a: a.issubset(customer_products))]

    recommendations = []
    for _, row in matches.sort_values('lift', ascending=False).iterrows():
        for product in row['consequents']:
            if product not in customer_products and product not in recommendations:
                recommendations.append(product)
        if len(recommendations) >= top_n:
            break

    return recommendations[:top_n]

sample_customer = df_exploded['Customer_Name'].iloc[0]
print(f"Recommendations for {sample_customer}:", recommend_products(sample_customer, df_exploded, segment_rules))
```

## 10. Product-Category Spend/Transaction Features → Customer Clustering
### Q: Do category-wise spending patterns reveal distinct customer clusters?

```python

# --- Step 1: Category-wise Spend ---
category_spend = df_exploded.pivot_table(
    index='Customer_Name', columns='Product_Category', values='Total_Cost', aggfunc='sum', fill_value=0
)
category_spend.columns = [f"{c}_Spend" for c in category_spend.columns]

# --- Step 2: Category-wise Transaction Count ---
category_txns = df_exploded.pivot_table(
    index='Customer_Name', columns='Product_Category', values='Transaction_ID', aggfunc='nunique', fill_value=0
)
category_txns.columns = [f"{c}_Txns" for c in category_txns.columns]

# --- Step 3: Category-wise Average Transaction Value ---
category_avg_value = category_spend.values / category_txns.replace(0, np.nan).values
category_avg_value = pd.DataFrame(
    category_avg_value, index=category_spend.index,
    columns=[c.replace('_Spend', '_AvgTxnValue') for c in category_spend.columns]
).fillna(0)

# --- Combine the three category-wise feature sets ---
category_features = pd.concat([category_spend, category_txns, category_avg_value], axis=1)

# --- Step 4: Bring in RFM ---
rfm_features = rfm.set_index('Customer_Name')[['Recency', 'Frequency', 'Monetary']]

# --- Merge category features with RFM ---
customer_features = category_features.join(rfm_features, how='inner')
customer_features.head()

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

scaler = StandardScaler()
X = scaler.fit_transform(customer_features)

import matplotlib.pyplot as plt

inertia = []
k_range = range(2, 11)

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X)
    inertia.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(list(k_range), inertia, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia (Within-Cluster Sum of Squares)')
plt.title('Elbow Method for Optimal k')
plt.xticks(list(k_range))
plt.grid(True, alpha=0.3)
plt.show()

for k, i in zip(k_range, inertia):
    print(f"k={k}: inertia={i:.1f}")
```

```python
chosen_k = 4

kmeans_final = KMeans(n_clusters=chosen_k, random_state=42, n_init=10)
customer_features['Cluster'] = kmeans_final.fit_predict(X)

print(f"=== Final Cluster Profiles (k={chosen_k}) ===")
print(customer_features.groupby('Cluster').mean().round(2))
print("\n=== Cluster Sizes ===")
print(customer_features['Cluster'].value_counts())
```

```python

centroids_scaled = kmeans_final.cluster_centers_
feature_names = customer_features.drop(columns='Cluster').columns.tolist()

n_features = len(feature_names)
angles = np.linspace(0, 2 * np.pi, n_features, endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(10, 10), subplot_kw=dict(polar=True))
colors = plt.cm.tab10.colors

for cluster_idx in range(chosen_k):
    values = centroids_scaled[cluster_idx].tolist()
    values += values[:1]
    ax.plot(angles, values, linewidth=2, label=f'Cluster {cluster_idx}', color=colors[cluster_idx])
    ax.fill(angles, values, alpha=0.1, color=colors[cluster_idx])

ax.set_xticks(angles[:-1])
ax.set_xticklabels(feature_names, fontsize=8)
ax.set_title(f'Cluster Profiles (k={chosen_k}) — Standardized Feature Values', y=1.08)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))

plt.tight_layout()
plt.show()
```

## 11. Product Recommendation (Market Basket Analysis)

```python
from mlxtend.frequent_patterns import apriori, association_rules

# --- Overall category-level basket (one row per transaction, one column per category) ---
category_basket = (
    df_exploded.groupby(['Transaction_ID', 'Product_Category'])['Product_Category']
    .count()
    .unstack(fill_value=0)
)
category_basket_encoded = category_basket.map(lambda x: 1 if x > 0 else 0)

frequent_category_itemsets = apriori(category_basket_encoded, min_support=0.01, use_colnames=True)
category_rules = association_rules(frequent_category_itemsets, metric="lift", min_threshold=1)
category_rules = category_rules.sort_values('lift', ascending=False)

print("=== Overall Category-Level Rules (lift > 1) ===")
print(f"Frequent category itemsets: {len(frequent_category_itemsets)}")
print(f"Rules with lift > 1: {len(category_rules)}")
print(category_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))

# --- Category-level rules by RFM segment ---
segment_category_rules = {}

for segment in df_exploded['Segment'].dropna().unique():
    segment_df = df_exploded[df_exploded['Segment'] == segment]
    basket = (
        segment_df.groupby(['Transaction_ID', 'Product_Category'])['Product_Category']
        .count()
        .unstack(fill_value=0)
    )
    basket_encoded = basket.map(lambda x: 1 if x > 0 else 0)

    try:
        frequent_itemsets = apriori(basket_encoded, min_support=0.01, use_colnames=True)
        rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
        segment_category_rules[segment] = rules.sort_values('lift', ascending=False)
    except Exception as e:
        print(f"Skipped {segment}: {e}")

for segment, rules in segment_category_rules.items():
    print(f"\n=== Top Category Rules for Segment: {segment} ===")
    print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(5))

categories_per_txn = (
    df_exploded.groupby(['Segment', 'Transaction_ID'])['Product_Category']
    .nunique()
    .rename('Distinct_Categories')
    .reset_index()
)

print("=== Avg Distinct Categories per Basket, by Segment ===")
print(categories_per_txn.groupby('Segment')['Distinct_Categories'].mean().sort_values(ascending=False).round(2))    
```
